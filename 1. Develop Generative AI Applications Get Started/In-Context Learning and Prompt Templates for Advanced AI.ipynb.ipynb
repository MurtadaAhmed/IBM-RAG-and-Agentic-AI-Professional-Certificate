```python
def llm_model(prompt_txt, params=None):
    
    model_id = "ibm/granite-3-8b-instruct"

    default_params = {
        "max_new_tokens": 256,
        "min_new_tokens": 0,
        "temperature": 0.5,
        "top_p": 0.2,
        "top_k": 1
    }

    url = "https://us-south.ml.cloud.ibm.com"
    project_id = "skills-network"
    
    granite_llm = WatsonxLLM(
        model_id=model_id,
        project_id=project_id,
        url=url,
        params=default_params
    )
    
    response = granite_llm.invoke(prompt_txt)
    return response
```

```python
template = """
tell me a {adjective} joke about {content}.
""" 
prompt = PromptTemplate.from_template(template)
prompt
```

```python
prompt.format(adjective='funny', content='chickens')
```

```python
from langchain_core.runnables import RunnableLambda
def format_template(variables):
    return prompt.format(**variables)
```

```python
joke_chain = (
    RunnableLambda(format_template)
    | llm
    | StrOutputParser()
)
joke_chain
```

```python
response = joke_chain.invoke(
    {"adjective": "funny", "content": "chickens"}
)
print(response)
```

```python
response = joke_chain.invoke({"adjective": "sad", "content": "fish"})
print(response)
```

```python
prompt = PromptTemplate.from_template(template)


def format_template(variables):
    return prompt.format(**variables)

role = """
    Dungeon & Dragons game master
"""

tone = "engaging and immersive"

template = """
You are an expert {role}. 
I have this question {question}.
I would like our conversation to be {tone}.
"""

roleplay_chain = (
    RunnableLambda(format_template)
    | llm
    | StrOutputParser()
)

while True:
    query = input("Question: ")

    if query.lower() in ["quit", "exit", "bye"]:
        print("Goodbye")
        break

    response = roleplay_chain.invoke(
        {"role": role, "question": query, "tone": tone}
    )

    print("Answer: ", response)
```
